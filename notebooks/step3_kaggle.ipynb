# 0. Pyroton V2 Kaggle ORPO pilot

Attach the private `pyroton-v2-inputs` dataset, enable Internet for pinned dependencies and model files,
and select a GPU accelerator. Training exposes only the first GPU and uses FP16.

Train or resume from cell 3. Each cell locates its files independently. For a fresh Kaggle VM,
attach the previous saved output under `/kaggle/input/pyroton-v2-resume` with its complete `checkpoints`
directory, or restore that directory from your saved outputs before running cell 3. An adapter ZIP
alone is suitable for evaluation and does not contain the optimizer state required for resume.

Use Save Version with Run All to retain the adapter ZIPs, checkpoints, and training metrics in Output.
Evaluation runs locally on Windows with the two released adapter ZIPs.


```python
# 1. Verify the sealed 89-pair dataset
from pathlib import Path
import subprocess
import sys
from zipfile import ZipFile

inputs = Path("/kaggle/input/pyroton-v2-inputs")
if not inputs.is_dir():
    matches = [path for path in Path("/kaggle/input/datasets").glob("*/pyroton-v2-inputs") if path.is_dir()]
    if len(matches) != 1:
        raise RuntimeError("Attach the private pyroton-v2-inputs dataset from the notebook Input panel.")
    inputs = matches[0]
working = Path("/kaggle/working")
working.mkdir(exist_ok=True)
bundle = inputs / "pyroton-v2-inputs.zip"
if bundle.is_file():
    target = working / "pyroton-inputs"
    target.mkdir(exist_ok=True)
    expected = {"kaggle_step3.py", "train_step3_pilot.py", "orpo-length-matched.jsonl",
                "quality-report.json", "stock-noise.json"}
    with ZipFile(bundle) as archive:
        if len(archive.namelist()) != len(expected) or set(archive.namelist()) != expected or archive.testzip():
            raise ValueError("The pilot ZIP does not contain the five sealed input files.")
        for name in sorted(expected):
            (target / name).write_bytes(archive.read(name))
    inputs = target

subprocess.run([sys.executable, str(inputs / "kaggle_step3.py"), "validate", "--inputs", str(inputs),
                "--working", str(working)], check=True)
```
